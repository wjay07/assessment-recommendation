```python
import pandas as pd

# Load the CSV file
csv_file = '/content/individual_test_solutions.csv'

# Read the CSV into a DataFrame
df = pd.read_csv(csv_file)

# Show the first few rows to verify the data
print(df.head())
```

```python
# Example: Test column has "A, C, P"
def expand_test_types(df):
    categories = {
        'A': 'Ability & Aptitude',
        'B': 'Biodata & Situational Judgement',
        'C': 'Competencies',
        'D': 'Development & 360',
        'E': 'Assessment Exercises',
        'K': 'Knowledge & Skills',
        'P': 'Personality & Behavior',
        'S': 'Simulations'
    }

    for key, name in categories.items():
        df[name] = df['test'].apply(lambda x: int(key in str(x)))
    return df
```

```python
print(df.columns.tolist())
```

```python
df = expand_test_types(df)
df.head()
```

```python
df['adaptive_irt'] = df['adaptive_irt'].replace('Unknown', 'No')
print(df['adaptive_irt'].value_counts())
```

```python

# Show column-wise null value counts
print("\n🔍 Null Value Count per Column:")
print(df.isnull().sum())

# Show total number of rows
print(f"\n📊 Total Rows: {len(df)}")

# Show unique values in each column
print("\n🔎 Unique Values per Column:")
for col in df.columns:
    print(f"- {col} ({df[col].dtype}):")
    print(df[col].unique()[:5])  # Show first 5 unique values only
    print()

# Show data types and non-null counts
print("\n🧾 DataFrame Info:")
print(df.info())

# Optional: Show descriptive statistics for numeric/time columns
print("\n📈 Descriptive Stats:")
print(df.describe(include='all'))
```

```python
# Fill missing values
df['languages'] = df['languages'].fillna('NA')
df['time'] = df['time'].fillna(0)  # Assuming 0 means untimed
df['download'] = df['download'].fillna('NA')

# Checking the result
print(df[['languages', 'time', 'download']].isnull().sum())
```

```python
# Print unique values and their counts for each column
for column in df.columns:
    print(f"Unique values and counts for '{column}':")
    print(df[column].value_counts())
    print("\n" + "-"*50 + "\n")
```

```python
# Replace unwanted time values with 0
df['time'] = df['time'].replace([
    'Approximate Completion Time in minutes = Untimed',
    'Approximate Completion Time in minutes = -',
    'Approximate Completion Time in minutes = Variable',
    'Approximate Completion Time in minutes = TBC',
    'Approximate Completion Time in minutes = N/A'
], 0)

# Verify if replacement is done
print(df['time'].value_counts())
```

```python
df['is_untimed'] = df['time'].apply(lambda x: 1 if x == 0 else 0)
```

```python
df['combined_description'] = df['title'] + ' ' + df['description'] + ' ' + df['languages'] + ' ' + df['test'] + ' Untimed: ' + df['is_untimed'].astype(str)
```
